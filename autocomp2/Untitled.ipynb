```python
import yaml
from dataclasses import dataclass, field
from typing import Optional

@dataclass
class ExperimentConfig:
    id: str
    model_name: str
    method: str
    timestamp: str
    train_data: str
    control_data: list[str]
    eval_data: list[str]
    detect_layers: list[int]
    control_dataset_size: int
    val_fraction: float
    reg_coeff: float
    control_on_policy: bool
    eval_on_policy: bool
    train_on_policy: bool
    normalize_acts: bool
    pw_locked: bool
    trim_reasoning: bool
    use_followup_question: bool
    use_goodfire_sae_acts: bool
    use_local_sae_acts: bool
    detect_only_last_token: bool
    detect_only_start_of_turn: bool
    lora_path: Optional[str] = None
    sae_latent_whitelist: Optional[str] = None
    detect_num_latents: Optional[int] = None
    max_llama_token_length: Optional[int] = None


def parse_config(path: str) -> ExperimentConfig:
    """Parse a YAML config file into an ExperimentConfig dataclass."""
    with open(path, "r") as f:
        raw = yaml.safe_load(f)

    return ExperimentConfig(
        id=raw["id"],
        model_name=raw["model_name"],
        method=raw["method"],
        timestamp=raw["timestamp"],
        train_data=raw["train_data"],
        control_data=raw.get("control_data", []),
        eval_data=raw.get("eval_data", []),
        detect_layers=raw.get("detect_layers", []),
        control_dataset_size=raw["control_dataset_size"],
        val_fraction=raw["val_fraction"],
        reg_coeff=raw["reg_coeff"],
        control_on_policy=raw["control_on_policy"],
        eval_on_policy=raw["eval_on_policy"],
        train_on_policy=raw["train_on_policy"],
        normalize_acts=raw["normalize_acts"],
        pw_locked=raw["pw_locked"],
        trim_reasoning=raw["trim_reasoning"],
        use_followup_question=raw["use_followup_question"],
        use_goodfire_sae_acts=raw["use_goodfire_sae_acts"],
        use_local_sae_acts=raw["use_local_sae_acts"],
        detect_only_last_token=raw["detect_only_last_token"],
        detect_only_start_of_turn=raw["detect_only_start_of_turn"],
        lora_path=raw.get("lora_path"),
        sae_latent_whitelist=raw.get("sae_latent_whitelist"),
        detect_num_latents=raw.get("detect_num_latents"),
        max_llama_token_length=raw.get("max_llama_token_length"),
    )
```
